```python
import pandas as pd

gene = pd.read_csv("/content/GSE160306_human_retina_DR_totalRNA_normalized_cpm.csv")
labels = pd.read_csv("/content/GSE160306_series_matrix1.csv")

print(gene.columns)
print(labels.columns)
```

```python
gene_samples = gene.columns[1:]   # ignore ensemblID column
label_samples = labels.columns

common_samples = list(set(gene_samples) & set(label_samples))

print(len(common_samples))
```

```python
gene_samples = gene.columns[1:]          # skip ensemblID
label_samples = labels.columns[1:]
common_samples = [s for s in gene_samples if s in label_samples]

gene_filtered = gene[["ensemblID"] + common_samples]
labels_filtered = labels[[labels.columns[0]] + common_samples]
```

```python
meta = labels_filtered.set_index(labels_filtered.columns[0]).T
meta.index.name = "sample_id"
meta.reset_index(inplace=True)
```

```python
if "!Sample_characteristics_ch1" in meta.columns:
    meta["label"] = meta["!Sample_characteristics_ch1"].apply(
        lambda x: "Control" if "Control" in str(x) else "DR"
    )
```

```python
rename_dict = {
    "!Sample_geo_accession": "GSM_id",
    "!Sample_title": "sample_title"
}
meta.rename(columns=rename_dict, inplace=True)
```

```python
gene_t = gene_filtered.set_index("ensemblID").T
gene_t.index.name = "sample_id"
gene_t.reset_index(inplace=True)
```

```python
GSE160306 = pd.merge(meta, gene_t, on="sample_id", how="inner")
```

```python
GSE160306.to_excel("/content/GSE160306_data.xlsx", index=False)
GSE160306.to_csv("/content/GSE160306_data.csv", index=False)

print(GSE160306.head())
print( GSE160306.shape)
```

```python
import pandas as pd
import mygene

# load file
df = pd.read_csv("/content/GSE160306.csv")

# first column is Sample, all other columns are ENSG IDs
ensembl_ids = df.columns[1:].tolist()

# remove version numbers if present
ensembl_ids = [x.split(".")[0] for x in ensembl_ids]

# query mygene
mg = mygene.MyGeneInfo()
results = mg.querymany(
    ensembl_ids,
    scopes="ensembl.gene",
    fields="symbol",
    species="human"
)

# build mapping
ensembl_to_symbol = {}
for r in results:
    ensg = r.get("query")
    symbol = r.get("symbol", ensg)  # keep ENSG if no symbol found
    ensembl_to_symbol[ensg] = symbol

# rename gene columns
new_columns = ["Sample"] + [ensembl_to_symbol.get(col.split(".")[0], col) for col in df.columns[1:]]
df.columns = new_columns

# save
df.to_csv("/content/GSE160306_with_gene_symbols.csv", index=False)

print(df.head())
```

```python
import pandas as pd
import mygene

df = pd.read_csv("/content/GSE160306_with_gene_symbols.csv")

ensg_cols = [c for c in df.columns if c.startswith("ENSG")]
print(len(ensg_cols))
```

```python
mg = mygene.MyGeneInfo()

results = mg.querymany(
    ensg_cols,
    scopes="ensembl.gene",
    fields="symbol",
    species="human"
)

mapping = {}

for r in results:
    query = r.get("query")
    symbol = r.get("symbol")

    if symbol:
        mapping[query] = symbol
    else:
        mapping[query] = query  # keep ENSG if no mapping
```

```python
df.rename(columns=mapping, inplace=True)

df.to_csv("/content/GSE160306_gene_symbols_updated.csv", index=False)
```
